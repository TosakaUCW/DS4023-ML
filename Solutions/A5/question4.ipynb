## **Steo-by-step calculation**:

```python
import numpy as np
import pandas as pd
from scipy.linalg import eig
```

```python
A = np.array([[1,2,3,4],[5,5,6,7],[1,4,2,3],[5,3,2,1],[8,1,2,2]]) #5 data points with 4 features
A
```

```python
df = pd.DataFrame(A,columns = ['f1', 'f2', 'f3','f4'])
df
```

```python
#Standardize the data points
df_std = df-df.mean()
df_std
```

```python
#calculate the covariance
df_cov = np.cov(df_std.T)
df_cov
```

```python
#eigenvalue decomposition
eig_values, eig_vectors = eig(df_cov)
```

```python
#eigenvectors
eig_vectors
```

```python
#eigenvalues
eig_values
```

```python
#Choose the first 2 eigenvectors
eig_2d = eig_vectors[:,:2]
eig_2d
```

```python
self_calculated = np.dot(df_std,eig_2d)
self_calculated
```

```python
# We can also directly import PCA function from sklearn to 
from sklearn.decomposition import PCA
pca = PCA(2)
reduced_data = pca.fit_transform(A)
```

```python
# Compare self-calculated result and result calculated by sklearn
print("Step-by-step calculated\n", self_calculated)
print("Sklearn calculated\n", reduced_data)
```
