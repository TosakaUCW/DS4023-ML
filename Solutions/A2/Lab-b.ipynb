# Linear Regression Lab Work

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn import datasets, linear_model
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
# Load the 2.1 Exercise dataset
dataset = pd.read_csv("2.2-Exercise.csv", header=0)
dataset
```

```python
# The diabetes dataset contains the following columns:
features_col = ['High']
target_col = ['Target']
```

```python
high_df = pd.DataFrame(dataset, columns=features_col)
target_df = pd.DataFrame(dataset, columns=target_col)
```

```python
# Split the data set into testing and training data
x_train, x_test, y_train, y_test = train_test_split(high_df, target_df, random_state=0)
```

```python
# Create a linear regression object
regression = linear_model.LinearRegression()
# Train the model using the training set
regression.fit(x_train, y_train)
# Make predictions using the testing set
y_predictions = regression.predict(x_test)
```

```python
# Plot the data
sns.set_style("darkgrid")
sns.regplot(x_test, y_test, fit_reg=False)
plt.plot(x_test, y_predictions, color='black')
# Remove ticks from the plot
plt.xticks([])
plt.yticks([])

plt.tight_layout()
plt.show()
```
